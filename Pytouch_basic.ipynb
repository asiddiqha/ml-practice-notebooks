```python
# Import PyTorch library
import torch

# Neural network module
import torch.nn as nn

# Optimizers like Adam, SGD
import torch.optim as optim

# Dataset and DataLoader utilities
from torch.utils.data import Dataset, DataLoader

# NumPy (used for array operations)
import numpy as np
```

```python
# Create a tensor from a Python list

x = torch.tensor([10, 20, 30, 40])
print(x)
```

What is it?

A Tensor is the basic data structure in PyTorch.

Python → List

NumPy → ndarray

PyTorch → Tensor

Everything in Deep Learning is stored as a tensor.

2. **Tensor Data Type**

```python
# Check the data type of a tensor

x = torch.tensor([1, 2, 3])
print(x.dtype)
```

**3. Tensor Shape**

```python
# Get the shape (rows, columns) of a tensor

x = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])
print(x.shape)
```

**4. Tensor Dimensions**

```python
# Find the number of dimensions

x = torch.tensor([[1, 2],
                  [3, 4]])
print(x.ndim)
```

Scalar      → 0D

Vector      → 1D

Matrix      → 2D

Image       → 3D

Video       → 4D

Batch Images→ 4D

**5. Create Different Tensors**

```python
# Create a tensor filled with zeros

x = torch.zeros((3, 4))
print(x)
```

```python
# Create a tensor filled with ones

x = torch.ones((2, 5))
print(x)
```

```python
# Create a tensor with random values

x = torch.rand((3, 3))
print(x)
```

Why?

Neural network weights are initialized randomly before training.

**6. Tensor Arithmetic**

Addition

```python
# Add two tensors element by element

a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])
print(a + b)
```

Subtraction

```python
# Subtract one tensor from another
print(a - b)
```

Multiplication

```python
# Multiply corresponding elements
print(a * b)
```

Division

```python
# Divide corresponding elements
print(a / b)
```

These are the mathematical operations performed throughout neural network training.

**7. Matrix Multiplication**

```python
# Matrix multiplication (not element-wise)

A = torch.tensor([[1, 2],
                  [3, 4]])

B = torch.tensor([[5, 6],
                  [7, 8]])

print(torch.matmul(A, B))
```

Why?

Every ANN layer performs matrix multiplication.

Output = Input × Weight + Bias

**8. Reshape Tensor**

```python
# Change tensor shape without changing data

x = torch.tensor([1, 2, 3, 4, 5, 6])
print(x.reshape(2, 3))
```

Used when preparing data for different neural network layers.

**9. NumPy to Tensor**

```python
# Convert a NumPy array to a PyTorch tensor
import numpy as np
import torch

arr = np.array([10, 20, 30])
tensor = torch.from_numpy(arr)
print(tensor)
```

Why?

Most preprocessing uses NumPy or Pandas. PyTorch models require tensors.

**10. Tensor to NumPy**

```python
# Convert a PyTorch tensor back to a NumPy array

x = torch.tensor([5, 10, 15])
print(x.numpy())
```

Useful for plotting, exporting, or using NumPy-based libraries.

**11. GPU Support**

```python
# Check whether a CUDA-enabled GPU is available

print(torch.cuda.is_available())
```

Move a tensor to GPU:

```python
# Move tensor to GPU if available

device = "cuda" if torch.cuda.is_available() else "cpu"
x = torch.tensor([1, 2, 3]).to(device)
print(x)
```

Why?

GPUs dramatically speed up deep learning training by performing many calculations in parallel.

**12. Automatic Gradient (Autograd)**

```python
# Compute gradients automatically

x = torch.tensor(5.0, requires_grad=True)
y = x**2
y.backward()
print(x.grad)
```

Why?

Backpropagation relies on gradients. PyTorch computes them automatically, so you don't have to derive them by hand.
